```python
def calculate_sum(number):
    return sum(number)

#define the number of list outside the function
number = [20,65,32,11,9]

#call the function with the list as argument
result = calculate_sum(number)
print("Input :",number)
print("Output :", result)
```

```python
#defining the function remove_duplicates
def remove_duplicates(number):
    
    return list(set(number))

#define the number of function with list as argument
number = [3,4,5,5,6,7,8,8]

#call the function with duplicate elements removed
result = remove_duplicates(number)
print("Input :", number)
print("Output :", result)
```

```python
#defining the function find_max
def find_max(number):
    return max(number)

#define the number of function with list as an argument
number = [20,80,5,9,6]

#call the function with duplicates removed
result = find_max(number)
print("Input :", number)
print("Output :", result)
```
