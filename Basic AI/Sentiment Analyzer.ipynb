```python
import openai

def get_response_from_chatgpt(text):
    prompt = f"Identify and return the sentiment (positive or negative) in the given text. text: {text}"
    
    response = openai.ChatCompletion.create(
        model='gpt-3.5-turbo',
        messages=[
            {"role": "system", "content": "You are a helpful sentiment analyzer that returns concise sentiment"},
            {"role": "user", "content": prompt}
        ],
        temperature=0.1
    )

    sentiment = response['choices'][0]['message']['content']
    return sentiment
```
