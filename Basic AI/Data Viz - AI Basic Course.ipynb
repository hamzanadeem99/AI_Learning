```python
df = pd.read_csv("titanic_dataset.csv")
df.head()
```

```python
df.shape
```

```python
sns.histplot(df['Parch'], kde = False)
```

```python
sns.distplot(df['Age'], hist = False)
```

```python
plt.figure(figsize = (8,8))
sns.distplot(df['Age'])
plt.show()
```

```python
sns.relplot(x = "Age", y = "Fare", col = "Pclass", hue = "Sex", kind = "line", data = df)
plt.show()
```

```python
plt.figure(figsize = (8,8))
sns.scatterplot(x = "Age", y = "Fare", hue = "Sex", data = df)
plt.show()
```

```python
plt.figure(figsize = (8,8))
sns.lineplot(x = "Age", y = "Fare", hue = "Sex", data = df)
plt.show()
```

```python
plt.figure(figsize = (8,8))
sns.barplot(x = "Sex", y = "Survived", hue = "Pclass", data = df)
plt.show()
```

```python
plt.figure(figsize = (8,8))
sns.stripplot(x = "Sex", y = "Age", hue = "Sex", data = df)
plt.show()
```

```python
plt.figure(figsize = (8,8))
sns.swarmplot(x = "Sex", y = "Age", hue = "Sex", data = df)
plt.show()
```

```python
plt.figure(figsize = (8,8))
sns.boxplot(x = "Survived", y = "Age", data = df)
plt.show()
```

```python
sns.violinplot(x = "Survived", y = "Age", hue = "Sex", data = df)
plt.show()
```

```python
sns.countplot(x = "Survived", data = df, palette = "Blues")
plt.show()
```

```python
plt.subplots(figsize = (8,8))
sns.pointplot(x = "Sex", y = "Survived", hue = "Pclass", data = df)
plt.show()
```

```python
sns.lmplot(x = "Age", y = "Fare", data = df)
plt.show()
```

```python
plt.subplots(figsize = (10,10))
sns.regplot(x = "Age", y = "Fare", data = df)
plt.show()
```

```python
plt.subplots(figsize = (10,10))
sns.regplot(x = "Fare", y = "Fare", data = df)
plt.show()
```

```python
plt.subplots(figsize=(10, 10))
sns.heatmap(df.corr(), cmap = "YlGnBu", annot=True, fmt=".2f")
plt.show()
```
