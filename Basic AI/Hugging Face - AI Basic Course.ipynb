```python
nlp = pipeline("ner")
example = "My name is Ahmad and i am going to Pakistan"

ner_results = nlp(example)
print(ner_results)
```

```python
nlp = pipeline("ner", model = "dslim/bert-base-NER")
example = "My name is Ahmad and i am going to Pakistan"

ner_results = nlp(example)
print(ner_results)
```
