```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import sklearn
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
plt.rcParams["figure.figsize"] = [10,5]
# Ignore warnings

import warnings

#Set the warning filter to ignore Future Warnings
warnings.simplefilter(action = "ignore", category = FutureWarning)
```

```python
full_data = pd.read_csv("titanic_dataset.csv")
```

```python
full_data.shape
```

```python
full_data.head()
```

```python
full_data.info()
```

```python
sns.heatmap(full_data.isnull(), yticklabels = False, cbar = False, cmap = "tab20c_r")
plt.title("Missing Data: Traning Set")
plt.show()
```

```python
plt.figure(figsize = (10,7))
sns.boxplot(x = "Pclass", y = "Age", data = full_data, palette = "GnBu_d").set_title("Age by Passenger Class")
plt.show()
```

```python
def impute_age(cols):
    Age = cols[0]
    Pclass = cols[1]
    
    if pd.isnull(Age):
    
        if pd.Pclass == 1:
            return 37
    
        elif Pclass == 2:
            return 29
    
        else:
            return 24
    
    else:
            return Age
# Apply the function to the Age column
full_data["Age"] = full_data[["Age", "Pclass"]].apply(impute_age, axis = 1)
```

```python
# Remove Cabin Feature
full_data.drop("Cabin", axis = 1, inplace = True)
```

```python
# Remove rows with missing data
full_data.dropna(inplace = True)
```

```python
full_data
```

```python
# Remove Unnecessary Columns
full_data.drop(["Name", "Ticket"], axis = 1, inplace = True)

# Convert objects to category data type
objcat = ["Sex", "Embarked"]

for colname in objcat:
    full_data[colname] = full_data[colname].astype("category")
```

```python
full_data
```

```python
full_data.describe()
```

```python
# Remove PassengerId
full_data.drop("PassengerId", inplace = True, axis = 1)
```

```python
full_data.shape
```

```python
# Identify categorical features
full_data.select_dtypes(['category']).columns
```

```python
# Convert categorical variables into "dummy" or indicator variables
sex = pd.get_dummies(full_data["Sex"], drop_first = True)

# drop_first prevents multi-collinearity
embarked = pd.get_dummies(full_data["Embarked"], drop_first = True)
```
